```python
import matplotlib.pyplot as plt

# X-axis data (Months)
months = [1, 4, 9, 16, 25, 36, 49, 64]

# Y-axis data for book sales
fiction_sales = [10, 25, 45, 60, 80, 95, 110, 130]
non_fiction_sales = [5, 15, 28, 40, 55, 70, 85, 100]

# canvas
plt.figure(figsize=(8, 6))

# Plot Fiction line (solid line with square markers)
plt.plot(months, fiction_sales, marker='s', color='y', label='Fiction')

# Plot Non-fiction line (dashed line with circle markers)
plt.plot(months, non_fiction_sales, linestyle='--', marker='o', color='g', label='Non-fiction')

# titles and axis labels
plt.title('Trend of Book Sales Over Time')
plt.xlabel('Months')
plt.ylabel('Number of Books Sold')

# legend in the lower right
plt.legend(loc='lower right')

plt.show()
```
