```python
import numpy as np
import pandas as pd
import math
import matplotlib.pyplot as plt

fig = plt.figure()
ax = fig.add_subplot(1, 1, 1)
ax.plot([90, 80, 70],[60, 70, 80])
plt.show()
```

```python
fig = plt.figure()
sales = fig.add_subplot(2, 1, 1)
profit = fig.add_subplot(2, 1, 2)
sales.plot(["pen", "paper", "ball", "bat", "wicket"],[1000, 1500, 2000, 2500, 3000])
profit.plot(["pen", "paper", "ball", "bat", "wicket"],[1000, 200, 500, 300, 500])
plt.tight_layout()
plt.show()
```

```python
fruits = ["pen", "paper", "table", "pencil", "Chair"]
sales1 = [1000, 2000, 3000, 4000, 5000]
profit1 = [1000, 2500, 3040, 460, 900]
plt.figure(figsize=(14, 7))
# First column
plt.subplot(1, 2, 1)
plt.plot(fruits, sales1)
plt.title("product_sales")
plt.xlabel("product_name")
plt.ylabel("sales")

#second colunm
plt.subplot(1, 2, 2)
plt.plot(fruits, profit1)
plt.title("products_profit")
plt.xlabel("product_name")
plt.ylabel("profit")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

xpoint = np.arange(0, 5*np.pi, 0.05)
ypoint = np.cos(xpoint)
plt.figure(figsize=(10, 5))
plt.plot(xpoint, ypoint, linewidth=2)
plt.title("cos wave")
plt.xlabel("Angle")
plt.ylabel("cos(x)")
plt.grid(True)
plt.show()
```
