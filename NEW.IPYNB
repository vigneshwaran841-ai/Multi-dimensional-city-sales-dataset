```python
import matplotlib.pyplot as plt

#example 1

fig = plt.figure()
ax = fig.add_subplot(1, 1, 1)
ax.plot([1, 2, 3], [4, 5, 6])

plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

fig = plt.figure()
ax1 = fig.add_subplot(2, 1, 1)
ax2 = fig.add_subplot(2, 1, 2)

ax1.plot([1, 2, 3], [4, 5, 6])
ax2.plot([1, 2, 3], [6, 5, 4])

plt.tight_layout()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([0, 1, 2, 3])

# Data
y1 = np.array([3, 8, 1, 10])
y2 = np.array([10, 20, 30, 40])

plt.figure(figsize=(12, 6))

# Plot 1
plt.subplot(1, 2, 1)
plt.plot(x, y1)
plt.title("Line-1")
plt.xlabel("x")
plt.ylabel("y1")

 # Plot 2
plt.subplot(1, 2, 2)
plt.plot(x, y2)
plt.title("Line-2")
plt.xlabel("x")
plt.ylabel("y2")

plt.tight_layout()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(0, 2 * np.pi, 0.05)
y = np.sin(x)

plt.figure(figsize=(10, 5))
plt.plot(x, y, linewidth=2)

plt.title("Sine Wave")
plt.xlabel("Angle (radians)")
plt.ylabel("sin(x)")

plt.grid(True)
plt.show()
```
