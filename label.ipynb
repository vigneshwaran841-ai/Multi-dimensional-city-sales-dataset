```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-3, 3, 30)
y = x**2

plt.figure(figsize=(8, 5))
plt.plot(x, y, label='$y = x^2$', linewidth=2)

plt.xlabel('x')
plt.ylabel('y')
plt.legend()

plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Create figure
plt.figure(figsize=(8, 6), dpi=80)  # Dots per inch (image quality)

# Subplot (only one plot, but keeping structure)
plt.subplot(1, 1, 1)

# Data
X = np.linspace(-np.pi, np.pi, 256)
C = np.cos(X)
S = np.sin(X)

# Plot curves
plt.plot(X, C, color="blue", linewidth=1, label="Cosine")
plt.plot(X, S, color="green", linewidth=1, label="Sine")

# Axis limits
plt.xlim(-4, 4)
plt.ylim(-1, 1)
```

```python
import matplotlib.pyplot as plt

year = [1972, 1982, 1992, 2002, 2012]
e_india = [100.6, 158.61, 305.54, 394.96, 724.79]
e_bangladesh = [10.5, 25.21, 58.65, 119.27, 274.87]

plt.figure(figsize=(9, 5))

plt.plot(year, e_india, color="orange", marker="o", linewidth=2, label="India")
plt.plot(year, e_bangladesh, color="green", marker="s", linewidth=2, label="Bangladesh")

plt.xlabel("Years")
plt.ylabel("Power consumption in kWh")
plt.title("Electricity Consumption per Capita: India vs Bangladesh")

plt.legend()
plt.grid(True)

plt.show()
```

```python
import matplotlib.pyplot as plt

# New data: Percentage of population using the Internet
year = [2000, 2005, 2010, 2015, 2020]

internet_india = [0.5, 2.4, 7.5, 27.0, 43.0]
internet_bangladesh = [0.07, 0.23, 3.7, 14.4, 24.8]

# Line plots
plt.plot(year, internet_india, color='orange', 
         marker='o', markersize=12, 
         label='India')

plt.plot(year, internet_bangladesh, color='g', 
         linestyle='dashed', linewidth=2, 
         label='Bangladesh')

plt.xlabel('Years')
plt.ylabel('Internet Users (% of population)')
plt.title('Growth of Internet Usage in India and Bangladesh (2000-2020)')

plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Create a Figure and multiple Axes objects in a grid layout
fig, axs = plt.subplots(2, 2, figsize=(10, 8))

# New Data
months = ['Jan', 'Feb', 'Mar', 'Apr']
sales = [120, 180, 260, 300]              # Line plot
profit = [20, 35, 45, 50]                 # Scatter plot
categories = ['Electronics', 'Clothing', 'Groceries', 'Furniture']
inventory = [40, 25, 60, 15]              # Bar plot
market_share = [35, 25, 20, 20]           # Pie chart

# Plot on individual Axes
axs[0, 0].plot(months, sales, marker='o')
axs[0, 1].scatter(months, profit)
axs[1, 0].bar(categories, inventory)
axs[1, 1].pie(market_share, labels=categories, autopct='%1.1f%%')

# Customize plot properties
axs[0, 0].set_title('Monthly Sales (in units)')
axs[0, 1].set_title('Monthly Profit (in thousands)')
axs[1, 0].set_title('Inventory Levels by Category')
axs[1, 1].set_title('Market Share Distribution')

# Improve layout
fig.tight_layout()

# Show the plot
plt.show()
```
