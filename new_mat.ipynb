```python
import numpy as np
import matplotlib.pyplot as plt

# Create figure
plt.figure(figsize=(8, 6), dpi=80)  #Dots per inch (image quality)

# Subplot (only one plot, but keeping structure)
plt.subplot(1, 1, 1)
 
# Data
X = np.linspace(-np.pi, np.pi, 200)
C = np.cos(X)
S = np.sin(X)

# Plot curves
plt.plot(X, C, color="blue", linewidth=1, label="Cosine")
plt.plot(X, S, color="green", linewidth=1, label="Sine")

# # Axis limits
# plt.xlim(-4, 4)
# plt.ylim(-1, 1)

# # Axis ticks
plt.xticks(np.linspace(-4, 4, 9))
plt.yticks(np.linspace(-1, 1, 5))

# Labels & legend
plt.xlabel("X")
plt.legend()
   
# Display plot
plt.show()
```

```python
import matplotlib.pyplot as plt

# Create a single figure with 3 rows and 2 columns
fig = plt.figure(figsize=(12, 12))

# ---------------------------------------------
# PART 1 — Using add_subplot()
# ---------------------------------------------

# Chart 1: Monthly Sales Growth (Line Plot)
ax1 = fig.add_subplot(3, 2, 1)
months = [1, 2, 3, 4, 5]
sales = [200, 260, 310, 370, 450]
ax1.plot(months, sales, marker='o', linewidth=2, color='blue')
ax1.set_xlabel('Month')
ax1.set_ylabel('Sales (units)')
ax1.set_title('Monthly Sales Growth')
ax1.grid(True)

# Chart 2: Revenue by Store (Bar Plot)
ax2 = fig.add_subplot(3, 2, 2)
stores = ['Store A', 'Store B', 'Store C', 'Store D']
revenue = [12000, 15000, 9000, 17000]
ax2.bar(stores, revenue, color='teal')
ax2.set_xlabel('Store')
ax2.set_ylabel('Revenue ($)')
ax2.set_title('Revenue by Store')

# ---------------------------------------------
# PART 2 — Using subplots() approach with add_subplot()
# ---------------------------------------------

# Chart 3: Ad Spend vs Sales (Scatter Plot)
ax3 = fig.add_subplot(3, 2, 3)
ad_spend = [10, 20, 30, 40]
sales_generated = [15, 35, 55, 80]
ax3.scatter(ad_spend, sales_generated, color='purple', s=100)
ax3.set_xlabel('Ad Spend ($)')
ax3.set_ylabel('Sales ($)')
ax3.set_title('Ad Spend vs Sales')
ax3.grid(True)

# Chart 4: Daily Website Traffic (Line Plot)
ax4 = fig.add_subplot(3, 2, 4)
days = [1, 2, 3, 4]
visitors = [400, 520, 610, 700]
ax4.plot(days, visitors, marker='s', color='green', linewidth=2)
ax4.set_xlabel('Day')
ax4.set_ylabel('Visitors')
ax4.set_title('Daily Website Traffic')
ax4.grid(True)

# Chart 5: Market Share by Category (Pie Chart)
ax5 = fig.add_subplot(3, 2, 5)
category_share = [40, 25, 20, 15]
labels = ['Electronics', 'Clothing', 'Home', 'Books']
colors = ['#ff9999', '#66b3ff', '#99ff99', '#ffcc99']
ax5.pie(category_share, labels=labels, autopct='%1.1f%%', colors=colors, startangle=90)
ax5.set_title('Market Share by Category')

# Chart 6: Customer Satisfaction (Line Plot)
ax6 = fig.add_subplot(3, 2, 6)
quarters = [1, 2, 3, 4]
ratings = [9.0, 8.5, 7.8, 7.0]
ax6.plot(quarters, ratings, marker='o', color='crimson', linewidth=2)
ax6.set_xlabel('Quarter')
ax6.set_ylabel('Rating')
ax6.set_title('Customer Satisfaction (Quarterly)')
ax6.grid(True)
ax6.set_ylim(6, 10)

# Adjust layout to prevent overlap
plt.tight_layout()

# Show the single figure with all charts
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# New dataset: Temperature variation (in °C) for two cities across 15 days
days = np.linspace(1, 15, 15)
cityA_temp = np.array([28, 30, 31, 33, 32, 34, 35, 36, 34, 33, 30, 29, 31, 32, 33])
cityB_temp = np.array([22, 24, 25, 27, 28, 29, 30, 31, 29, 28, 27, 26, 25, 24, 23])

# Manual axes creation [left, bottom, width, height]
ax = plt.axes([0.1, 0.1, 0.8, 0.8])

# Plotting data
ax.plot(days, cityA_temp, 'rx:', label="City A Temperature")
ax.plot(days, cityB_temp, 'cD-', label="City B Temperature")

# Title and legend
ax.set_title("Temperature Trends Across 15 Days")
ax.set_xlabel("Days")
ax.set_ylabel("Temperature (°C)")
ax.legend(loc="upper left")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

fig = plt.figure(figsize=(6, 6))

ax1 = plt.subplot2grid(shape=(3, 3), loc=(0, 0), colspan=3)   # Visits
ax2 = plt.subplot2grid(shape=(3, 3), loc=(1, 0), colspan=1)   # Bounce rate
ax3 = plt.subplot2grid(shape=(3, 3), loc=(1, 2), rowspan=2)   # Active users
ax4 = plt.subplot2grid((3, 3), (2, 0))                        # Conversions
ax5 = plt.subplot2grid((3, 3), (2, 1), colspan=1)             # Session duration

# New dataset: 0–30 days
x = np.arange(0, 30, 0.1)  

# Fake analytics data
visits = 500 + 150*np.sin(0.2 * x)          # ax1
bounce_rate = 40 + 10*np.cos(0.15 * x)      # ax2
active_users = 300 + 80*np.sin(0.1 * x)     # ax3
conversions = 20 + 5*np.cos(0.3 * x)        # ax4
session_time = 3 + 1*np.sin(0.25 * x)       # ax5

# Plotting subplots
ax1.plot(x, visits, color="blue")
ax1.set_title("Daily Visits")

ax2.plot(x, bounce_rate, color="orange")
ax2.set_title("Bounce Rate (%)")

ax3.plot(x, active_users, color="green")
ax3.set_title("Active Users")

ax4.plot(x, conversions, color="red")
ax4.set_title("Conversions (%)")

ax5.plot(x, session_time, color="purple")
ax5.set_title("Avg Session Duration (min)")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

N = 5
# New dataset: study hours for 5 groups of students
online_hours  = (10, 18, 15, 20, 12)
offline_hours = (12, 10, 14,  8, 11)

ind = np.arange(N)   # group positions
width = 0.35

fig = plt.figure()
ax = fig.add_axes([0, 0, 1, 1])

# Stacked bar chart
ax.bar(ind, online_hours, width, color='r')
ax.bar(ind, offline_hours, width, bottom=online_hours, color='g')

ax.set_ylabel('Study Hours')
ax.set_title('Online vs Offline Study Hours by Student Group')

ax.set_xticks(ind, ('G1', 'G2', 'G3', 'G4', 'G5'))
ax.set_yticks(np.arange(0, 41, 5))

ax.legend(labels=['Online Study', 'Offline Study'])

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

x = np.arange(1, 11)
y = np.random.randint(5, 20, 10)

plt.figure(figsize=(7, 4))
sns.lineplot(x=x, y=y)
plt.title("Single Line Plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

```python
sns.set_theme(context="talk", style="darkgrid", palette="coolwarm", font_scale=1.1)
# paper -> Small text, small lines
# notebook -> Normal size
# talk -> Large text, thick lines
# poster -> Very large text, very thick lines

# darkgrid -> Dark background, grid lines
# whitegrid -> White background, grid lines
# white -> White background, no grid
# ticks -> White background, ticks

plt.figure(figsize=(7, 4))
sns.lineplot(x=x, y=y)
plt.title("Styled Line Plot")
plt.show()

# Reset style
sns.set()
```
